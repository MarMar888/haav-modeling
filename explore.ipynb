# Boat model explorer

A working model of the boat from `optimization2.md`, for poking at. Change an input, re-run, and see what
moves. Units are imperial (ft, in, lb, mph, Wh) because that is how the design is discussed.

**Full write-up:** every decision variable, calculated value, constraint and the objective of this model are documented in a LaTeX overview: [read the PDF](https://github.com/MarMar888/hav/blob/master/optimization/explore-model.pdf) ([LaTeX source](https://github.com/MarMar888/hav/blob/master/optimization/explore-model.tex)).

**What is real and what is not**

| Piece | Status |
| --- | --- |
| Planing equilibrium (trim, wetted length, thrust needed) | Real: [OpenPlaning](https://github.com/elcf/python-openplaning), Savitsky's empirical method. Prismatic hull only. |
| Hydrostatics, stability, freeboard, hull mass | Real mesh hydrostatics: [navaltoolbox](https://github.com/NavalToolbox/navaltoolbox-lib) on a hull mesh the notebook generates (V-bottom, vertical sides, a bow that tapers to the stem). Draft, freeboard, metacentric height (GM) and the at-rest trim come from the mesh. The **bow taper fraction is a guess**, and the hull skin mass is the mesh area times a guessed laminate weight. |
| Propeller rpm, tip speed, Kv | Your equation: `rpm = mph x 1056 / (pitch x 0.85)`, `Kv = rpm / loaded voltage`. |
| Motor | Real data: the SSS brushless motors from the supplier spec sheet (`sss_motors.csv`, 263 windings). Each design picks one winding of one motor; Kv, max current, max voltage, resistance, no-load current, continuous power and weight come from the sheet. Efficiency and current come from the DC motor model at the propeller's operating point (`speed = Kv (V - I R)`, `torque = Kt (I - I0)`). |
| Propeller efficiency and mass | Ours: ideal actuator-disk (momentum) theory, so a bigger diameter is more efficient at the same thrust, with placeholder blade losses and a pitch/diameter penalty. Mass scales with diameter cubed. Tip speed is limited. |
| Battery voltage under load | Ours: `V = (Voc + sqrt(Voc^2 - 4 R P)) / 2`. |
| Cell data, speed controller efficiency, propeller blade losses and mass, current limit | **Placeholders**. Every one is marked `PLACEHOLDER` below. |

**Modes.** One cell at the top of section 2 holds the explicit switches: where the motor's limits come from (`MOTOR_LIMITS`), whether Savitsky's validity limits are enforced (`ENFORCE_VALIDITY`), whether doubtful windings join the catalog (`INCLUDE_SUSPECT_R`), and the course assumption (`COURSE_MODE`). The problem card in section 9 echoes them, so any chart can be traced back to its settings.

Nothing here is validated against a real boat. Use it to learn which variables matter and where the
constraints bite, not to read off a design.

**Version:** the first code cell prints the model version (`optimization/VERSION`, plus the git commit and whether
there are uncommitted edits). Check it before trusting a saved chart; see `docs/VERSIONING.md`.

**Run it:** `optimization/.venv/bin/jupyter lab optimization/explore.ipynb`, then Run All. The optimizer

```python
import csv, io, urllib.request


@dataclass(frozen=True)
class MotorSpec:
    # one winding of one SSS motor, from the supplier spec sheet (optimization/sss_motors.csv)
    name: str          # e.g. "SSS 56154-V2 5Y"
    kv: float          # rpm per volt
    i_max_a: float     # max current
    v_max_v: float     # max voltage
    r_ohm: float       # winding resistance (the sheet's resistance column, taken as ohms)
    i0_a: float        # no-load current; the sheet measures it at 11.1 V and it is held constant here
    p_cont_w: float    # continuous power rating of the motor
    p_max_w: float
    max_rpm: float
    mass_lb: float
    r_suspect: bool    # resistance entry looks like a typo on the sheet


CSV_URL = "https://raw.githubusercontent.com/MarMar888/hav/master/optimization/sss_motors.csv"


def load_motors():
    here = Path.cwd()
    local = next((p for p in (here / "sss_motors.csv", here / "optimization" / "sss_motors.csv") if p.exists()), None)
    text = local.read_text() if local else urllib.request.urlopen(CSV_URL, timeout=30).read().decode()
    out = []
    for r in csv.DictReader(io.StringIO(text)):
        i_max, v_max, res = float(r["i_max_a"]), float(r["v_max_v"]), float(r["r_ohm"])
        out.append(MotorSpec(f"{r['motor']} {r['winding']}", float(r["kv_rpm_per_v"]), i_max, v_max, res, float(r["i0_a_at_11v"]),
                             float(r["p_cont_w"]), float(r["p_max_w"]), float(r["max_rpm"]), float(r["weight_g"]) / 453.592,
                             not 0.01 <= i_max * res / v_max <= 0.30))
    return sorted(out, key=lambda s: (s.p_cont_w, s.name.split()[1], -s.kv))     # smallest motor first, windings of one motor together


ALL_MOTORS = load_motors()
CATALOG = [s for s in ALL_MOTORS if INCLUDE_SUSPECT_R or not s.r_suspect]


def motor_index(name):
    return next(i for i, s in enumerate(CATALOG) if s.name == name)


print(f"{len(CATALOG)} SSS windings in the catalog ({len(ALL_MOTORS) - len(CATALOG)} left out for a doubtful resistance entry)")
```

## 3. Hull: mesh hydrostatics and mass

The hull is a V-bottom section of chine beam `b` and deadrise `beta` with vertical sides up to the hull depth,
prismatic from the transom forward, with the chine beam tapering linearly to the stem over the forward
`bow_fraction` of the length. The notebook builds a watertight triangle mesh of that hull and hands it to
[navaltoolbox](https://github.com/NavalToolbox/navaltoolbox-lib), which finds the draft at which the displaced
water weighs as much as the boat and reports freeboard, the metacentric height GM (static roll stiffness) and
the at-rest trim from the same geometry. The hull skin mass is the mesh area times the laminate weight - the
same mesh that floats the boat also weighs it. Freeboard is measured at rest, which is conservative because
the boat rides higher when planing. `evaluate()` uses the level-trim state (fast, cached); the free-trim
equilibrium is shown in the overview and the problem card.

```python
import struct, tempfile
from functools import lru_cache
from navaltoolbox import Hull, Vessel, HydrostaticsCalculator


def chine_height_ft(d):
    return d.beam_in / 12 / 2 * math.tan(math.radians(d.deadrise_deg))


def hull_sections(length_ft, beam_in, deadrise_deg, depth_in, bow_fraction, nx_bow=8):
    # stations along the hull in SI: x transom -> bow, y to starboard, z up from the keel.
    # Prismatic aft body; over the forward bow_fraction the chine half-beam tapers linearly to the stem.
    L, B, D = length_ft * FT, beam_in * IN, depth_in * IN
    t = math.tan(math.radians(deadrise_deg))
    x_taper = L * (1.0 - bow_fraction)
    xs = [0.0] + ([x_taper] if x_taper > 0 else []) + [x_taper + (L - x_taper) * i / nx_bow for i in range(1, nx_bow + 1)]
    secs = []
    for x in xs:
        hb = B / 2 if x <= x_taper else B / 2 * (L - x) / (L - x_taper)
        secs.append([(x, 0.0, 0.0), (x, hb, hb * t), (x, hb, D), (x, 0.0, D)])   # keel, chine, deck edge, deck centre
    return secs


def hull_mesh(length_ft, beam_in, deadrise_deg, depth_in, bow_fraction):
    # watertight triangle mesh with outward normals (checked by the signed-volume assertion)
    secs = hull_sections(length_ft, beam_in, deadrise_deg, depth_in, bow_fraction)
    tris = []
    quad = lambda a, b, c, d_: tris.extend([(a, b, c), (a, c, d_)])
    for s0, s1 in zip(secs, secs[1:]):
        for k in range(3):                                   # bottom, side, half-deck (starboard)
            quad(s0[k], s0[k + 1], s1[k + 1], s1[k])
    mir = lambda p: (p[0], -p[1], p[2])
    for a, b, c in list(tris):                               # mirror to port, winding reversed
        tris.append((mir(a), mir(c), mir(b)))
    for sec, flip in ((secs[0], False), (secs[-1], True)):   # close the transom (outward -x) and the stem (outward +x)
        ring = [sec[0], sec[1], sec[2], sec[3], mir(sec[2]), mir(sec[1])]
        for i in range(1, len(ring) - 1):
            tris.append((ring[0], ring[i], ring[i + 1]) if flip else (ring[0], ring[i + 1], ring[i]))
    vol6 = sum(a[0] * (b[1] * c[2] - b[2] * c[1]) - a[1] * (b[0] * c[2] - b[2] * c[0]) + a[2] * (b[0] * c[1] - b[1] * c[0])
               for a, b, c in tris)
    assert vol6 > 0, "hull mesh is inside out"
    return tris


def _mesh_area_m2(tris):
    area = 0.0
    for a, b, c in tris:
        u = (b[0] - a[0], b[1] - a[1], b[2] - a[2]); v = (c[0] - a[0], c[1] - a[1], c[2] - a[2])
        n = (u[1] * v[2] - u[2] * v[1], u[2] * v[0] - u[0] * v[2], u[0] * v[1] - u[1] * v[0])
        area += math.sqrt(n[0] ** 2 + n[1] ** 2 + n[2] ** 2) / 2
    return area


_MESH_DIR = Path(tempfile.mkdtemp(prefix="hull-mesh-"))


def _write_stl(tris, path):
    with open(path, "wb") as fh:
        fh.write(b"\0" * 80); fh.write(struct.pack("<I", len(tris)))
        for a, b, c in tris:
            fh.write(struct.pack("<3f", 0.0, 0.0, 0.0))      # the loader recomputes normals from the winding
            for p in (a, b, c): fh.write(struct.pack("<3f", *p))
            fh.write(struct.pack("<H", 0))


@lru_cache(maxsize=4096)
def _hydro_calc(geom, rho):
    path = _MESH_DIR / f"{abs(hash((geom, rho))):x}.stl"
    _write_stl(hull_mesh(*geom), path)
    return HydrostaticsCalculator(Vessel.from_hulls([Hull(str(path))]), rho)


@lru_cache(maxsize=200_000)
def _hydro_state(geom, rho, mass_kg, vcg_m, lcg_m):
    try:
        if lcg_m is None:
            return _hydro_calc(geom, rho).from_displacement(mass_kg, vcg=vcg_m)
        return _hydro_calc(geom, rho).from_displacement(mass_kg, cog=(lcg_m, 0.0, vcg_m))
    except Exception:
        return None                                          # no equilibrium: the hull cannot displace this weight


def hydrostatics(d, f, mass_lb, vcg_m, lcg_ft=None, free_trim=False):
    """Draft, freeboard, GM and trim from navaltoolbox on the hull mesh. Level trim unless free_trim,
    which finds the pitch equilibrium for the given LCG (slower; used for charts, not in evaluate())."""
    geom = (round(d.length_ft, 4), round(d.beam_in, 4), round(d.deadrise_deg, 4), round(d.depth_in, 4), round(f.bow_fraction, 4))
    rho = round(f.water_lb_ft3 * LB_FT3_TO_KG_M3, 3)
    st = _hydro_state(geom, rho, round(mass_lb * LB_KG, 4), round(vcg_m, 4),
                      round(lcg_ft * FT, 4) if free_trim else None)
    if st is None or not 0.0 <= st.draft <= d.depth_in * IN + 1e-6:   # sunk, or the solver left its range
        return dict(ok=False, draft_in=d.depth_in, freeboard_in=0.0, gmt_m=-1.0, trim_deg=0.0, waterplane_ft2=float("nan"))
    return dict(ok=True, draft_in=st.draft / IN, freeboard_in=d.depth_in - st.draft / IN, gmt_m=st.gmt,
                trim_deg=st.trim, waterplane_ft2=st.waterplane_area / FT ** 2)


def shell_area_ft2(d, f):
    return _mesh_area_m2(hull_mesh(d.length_ft, d.beam_in, d.deadrise_deg, d.depth_in, f.bow_fraction)) / FT ** 2


def hull_mass_lb(d, f):
    return shell_area_ft2(d, f) * f.laminate_lb_ft2 + f.hull_structure_lb
```

## 4. Battery

`capacity = S x P x cell Ah x cell nominal volts`. The start-of-race charge is **derived**: the highest charge at which the
open-circuit voltage stays at or under the 55.5 V rule (capped at 100%). Under load the voltage sags by the pack's internal
resistance: solving `V = Voc - I R` with `I = P / V` gives `V = (Voc + sqrt(Voc^2 - 4 R P)) / 2`, which only has a solution
while `Voc^2 >= 4 R P`.

```python
def ocv(soc, cell):
    return float(np.interp(soc, cell.soc, cell.ocv))


def start_soc(series, f):
    cell = f.cell
    if series * cell.v_full <= f.voltage_limit_v:
        return 1.0
    return float(np.interp(f.voltage_limit_v / series, cell.ocv, cell.soc))


def pack(d, f):
    c = f.cell
    n = d.series * d.parallel
    return dict(cells=n, capacity_wh=n * c.ah * c.v_nominal,
                mass_lb=n * c.mass_lb * (1 + c.pack_overhead),
                resistance_ohm=d.series / d.parallel * c.resistance_ohm,
                current_limit_a=d.parallel * c.max_current_a)


def loaded_voltage(v_oc, r_pack, p_bus):
    disc = v_oc ** 2 - 4 * r_pack * p_bus
    return (v_oc + math.sqrt(disc)) / 2 if disc >= 0 else float("nan")
```

## 5. Motor and propeller

The motor is a real **SSS winding** chosen from `CATALOG` (`Design.motor`). Its mass is the weight on the supplier sheet, so a bigger motor
costs boat weight. At the propeller's operating point the DC motor model gives the current, the voltage and the efficiency:

- `Kt = 60 / (2 pi Kv)` (N m per amp), `torque = shaft power / angular speed`, `current I = torque / Kt + I0`.
- `voltage needed = rpm / Kv + I R`: the back-voltage plus the drop across the winding resistance.
- `electrical power = voltage needed x I`, so `motor efficiency = shaft power / electrical power`. A speed controller (`drive_eff`) sits
  between the battery bus and the motor.

Three limits come from the sheet: the current must stay under the winding's max current, the shaft power under its continuous rating, and the
full-charge pack voltage under its max voltage. A fourth is new: the loaded pack voltage must be at least the voltage the motor needs at that rpm.
Direct drive, no gearbox, so motor rpm is the propeller rpm below.

The propeller has a small model of its own, so diameter and pitch now matter:

- **Efficiency** starts from ideal actuator-disk (momentum) theory, `eta = 2 / (1 + sqrt(1 + Ct))` with `Ct = T / (0.5 rho A V^2)`.
  At the same thrust a bigger diameter means a more lightly loaded disk and a higher efficiency. Two placeholder factors sit on top:
  blade losses (`prop_blade_eff`) and a penalty for a pitch-to-diameter ratio away from `prop_pd_opt`.
- **rpm** is still your equation, `rpm = mph x 1056 / (pitch x 0.85)`, so a coarser pitch turns slower, and so asks the motor for more torque.
- **Tip speed** is `pi x diameter x rpm`, and it is limited (`max_tip_speed_fps`), which is what stops the diameter growing forever.
- **Mass** scales with diameter cubed (`prop_mass_lb`), so a bigger propeller also weighs more, sits at the transom, and moves the
  center of gravity aft.
- **Fit**: the propeller must be no deeper than the hull (a geometry constraint, `prop_fits`).

```python
def motor_caps(s):
    """Current and voltage cap for a winding under MOTOR_LIMITS. "sheet" takes the spec sheet's maxima;
    "theoretical" derives them from the DC model: back-EMF voltage at the sheet's rpm ceiling, and the
    current that carries the continuous power at that voltage."""
    if MOTOR_LIMITS == "sheet":
        return s.i_max_a, s.v_max_v
    v_cap = s.max_rpm / s.kv
    return s.p_cont_w / v_cap + s.i0_a, v_cap


def motor_spec(d):
    return CATALOG[d.motor]


def motor_mass_lb(d):
    return motor_spec(d).mass_lb                 # weight from the supplier sheet


def motor_point(d, shaft_w, rpm):
    # DC motor at one operating point: current, voltage needed and efficiency to put shaft_w on the shaft at rpm (direct drive)
    s = motor_spec(d)
    omega = rpm * 2 * math.pi / 60
    torque = shaft_w / omega
    kt = 60 / (2 * math.pi * s.kv)                       # N m per amp
    current = torque / kt + s.i0_a
    volts = rpm / s.kv + current * s.r_ohm               # back-voltage plus the drop across the winding
    elec_w = volts * current
    return dict(torque_nm=torque, current_a=current, volts_needed=volts, elec_w=elec_w, efficiency=shaft_w / elec_w)


def prop_mass_lb(d, f):
    # PLACEHOLDER: blade and hub volume scale with diameter cubed
    return f.prop_mass_ref_lb * (d.prop_diam_in / f.prop_ref_diam_in) ** 3


def prop_efficiency(d, f, thrust_n, speed_ms):
    # Ideal actuator disk (Froude): eta = 2 / (1 + sqrt(1 + Ct)), Ct = T / (0.5 rho A V^2). A lightly loaded disk (large diameter,
    # low thrust, high speed) is more efficient. Blade losses and a pitch/diameter penalty are placeholders on top.
    area = math.pi / 4 * (d.prop_diam_in * IN) ** 2
    ct = thrust_n / (0.5 * f.water_lb_ft3 * LB_FT3_TO_KG_M3 * area * speed_ms ** 2)
    eta_ideal = 2 / (1 + math.sqrt(1 + ct))
    pd = d.pitch_in / d.prop_diam_in
    eta_pd = max(0.3, 1 - f.prop_pd_penalty * (pd - f.prop_pd_opt) ** 2)
    return eta_ideal * f.prop_blade_eff * eta_pd


def prop_numbers(d, f, speed_mph, v_loaded):
    rpm = speed_mph * 1056 / (d.pitch_in * f.slip_factor)
    tip_fps = math.pi * d.prop_diam_in / 12 * rpm / 60
    return dict(rpm=rpm, tip_speed_fps=tip_fps, kv_needed=rpm / v_loaded if v_loaded == v_loaded else float("nan"))
```

## 6. Planing equilibrium (OpenPlaning)

For a given weight, LCG, VCG, speed, beam and deadrise, OpenPlaning finds the trim and wetted length where lift balances
weight and moments balance, and returns the thrust the propeller must supply. If it cannot find a solution the design is
treated as infeasible.

We also record Savitsky's applicability limits so we can enforce them (or not; see `ENFORCE_VALIDITY` below):
running trim 2 to 15 degrees, mean wetted length over beam at most 4, speed coefficient `Cv` 0.6 to 13, and the wetted
keel no longer than the hull.

```python
def planing(d, f, mass_lb, lcg_ft, vcg_m, speed_mph):
    # OpenPlaning warns on every call outside a method's range, and turns warnings back on internally
    # after each solve, so silence them for the duration of the call. We check the limits ourselves.
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        return _planing(d, f, mass_lb, lcg_ft, vcg_m, speed_mph)


def _planing(d, f, mass_lb, lcg_ft, vcg_m, speed_mph):
    try:
        b = PlaningBoat(speed=speed_mph * MPH, weight=mass_lb * LB_N, beam=d.beam_in * IN, lcg=lcg_ft * FT, vcg=vcg_m,
                        r_g=f.gyradius_frac * d.length_ft * FT, beta=d.deadrise_deg, epsilon=d.thrust_angle_deg,
                        vT=f.thrust_height_in * IN, lT=0.0, loa=d.length_ft * FT, H_sig=f.wave_height_in * IN,
                        rho=f.water_lb_ft3 * LB_FT3_TO_KG_M3, nu=f.water_nu)
        b.get_steady_trim()
        warnings.simplefilter("ignore")        # get_steady_trim re-enables warnings when it finishes
        b.get_forces()
        thrust = abs(float(b.thrust_force[0]))
        out = dict(ok=bool(np.isfinite(thrust)), trim_deg=float(b.tau), thrust_n=thrust, keel_m=float(b.L_K),
                   chine_m=float(b.L_C), lam=float(b.lambda_W), transom_draft_m=float(b.T))
        out["cv"] = speed_mph * MPH / math.sqrt(9.80665 * d.beam_in * IN)
        return out, b
    except Exception as e:                      # OpenPlaning raises when no equilibrium exists
        return dict(ok=False, error=str(e).splitlines()[0][:80]), None
```

## 7. Whole-boat evaluation

`evaluate(design, fixed)` walks the chain: hull, mass and balance, hydrostatics, planing thrust, propeller, power, battery,
race time, and then every constraint. Each constraint is a **margin** (positive passes) with a scale so violations can be
compared, grouped as `rule`, `limit`, `geometry` or `validity`.

Race time uses speed over the ground: `speed - current`. The course allowance is charged at cruise power.

```python
# ENFORCE_VALIDITY lives in the Modes cell in section 2.
@dataclass
class Result:
    m: dict                 # metrics
    margins: dict           # name -> (margin, scale, group)

    def feasible(self, enforce_validity=None):
        ev = ENFORCE_VALIDITY if enforce_validity is None else enforce_validity
        return all(v >= 0 for v, s, g in self.margins.values() if ev or g != "validity")

    def violation(self, enforce_validity=None):
        ev = ENFORCE_VALIDITY if enforce_validity is None else enforce_validity
        return sum(max(0.0, -v / s) for v, s, g in self.margins.values() if ev or g != "validity")

    def failing(self, enforce_validity=None):
        ev = ENFORCE_VALIDITY if enforce_validity is None else enforce_validity
        return [k for k, (v, s, g) in self.margins.items() if v < 0 and (ev or g != "validity")]


def evaluate(d, f=FIXED, speed_mph=None):
    v = d.speed_mph if speed_mph is None else speed_mph
    cell = f.cell
    m, g = {}, {}

    # hull, masses, balance
    hull_lb, mot_lb, bat = hull_mass_lb(d, f), motor_mass_lb(d), pack(d, f)
    parts = [  # name, lb, x as fraction of length, z in above keel
        ("hull", hull_lb, 0.45, f.hull_z_frac_of_depth * d.depth_in),
        ("payload", f.payload_lb, d.payload_x, f.payload_z_in),
        ("battery", bat["mass_lb"], d.battery_x, f.battery_z_in),
        ("motor", mot_lb, 0.10, f.motor_z_in),
        ("prop", prop_mass_lb(d, f), 0.0, f.motor_z_in),
        ("fixed", f.fixed_equipment_lb, 0.40, f.fixed_z_in),
    ]
    mass = sum(p[1] for p in parts)
    lcg_ft = sum(p[1] * p[2] for p in parts) / mass * d.length_ft
    vcg_m = sum(p[1] * p[3] for p in parts) / mass * IN
    m.update(mass_lb=mass, hull_lb=hull_lb, motor_lb=mot_lb, battery_lb=bat["mass_lb"],
             lcg_frac=lcg_ft / d.length_ft, vcg_m=vcg_m, capacity_wh=bat["capacity_wh"])

    # hydrostatics: navaltoolbox on the hull mesh, level trim (the free-trim state is a chart, not a constraint)
    hyd = hydrostatics(d, f, mass, vcg_m)
    wave_in = f.wave_allowance_factor * f.wave_height_in
    m.update(static_draft_in=hyd["draft_in"], freeboard_in=hyd["freeboard_in"], gmt_m=hyd["gmt_m"],
             waterplane_ft2=hyd["waterplane_ft2"])
    g["freeboard"] = (hyd["freeboard_in"] - f.freeboard_rule_in - wave_in, 3.0, "rule")
    g["stability"] = (hyd["gmt_m"] - f.min_gmt_m, 0.1, "limit")
    g["depth_over_chine"] = (d.depth_in - chine_height_ft(d) * 12, 12.0, "geometry")
    g["prop_fits"] = (d.depth_in - d.prop_diam_in, d.depth_in, "geometry")

    # planing thrust at this speed through the water
    p, boat = planing(d, f, mass, lcg_ft, vcg_m, v)
    m["planing"] = p
    ground = v - f.current_mph
    g["ground_speed"] = (ground, 5.0, "rule")
    if f.course_mode == "worst":
        hours = f.course_miles / ground if ground > 0 else float("inf")
    else:
        hours = (f.course_miles / 2 / (v - f.current_mph) + f.course_miles / 2 / (v + f.current_mph)) if ground > 0 else float("inf")
    race_s = hours * 3600 + f.allowance_s
    m["race_time_s"] = race_s

    g["solver_ok"] = (1.0 if p["ok"] else -1.0, 1.0, "validity")
    if not p["ok"]:
        return Result(m, g)

    # validity limits of the planing method
    g["trim_min"] = (p["trim_deg"] - 2.0, 2.0, "validity")
    g["trim_max"] = (15.0 - p["trim_deg"], 15.0, "validity")
    g["wetted_length_ratio"] = (4.0 - p["lam"], 4.0, "validity")
    g["cv_min"] = (p["cv"] - 0.6, 0.6, "validity")
    g["cv_max"] = (13.0 - p["cv"], 13.0, "validity")
    g["keel_within_hull"] = (d.length_ft * FT - p["keel_m"], d.length_ft * FT, "validity")

    # power chain
    eta_p = prop_efficiency(d, f, p["thrust_n"], v * MPH)
    shaft_w = p["thrust_n"] * v * MPH / eta_p
    rpm = v * 1056 / (d.pitch_in * f.slip_factor)
    mot = motor_point(d, shaft_w, rpm)
    bus_w = mot["elec_w"] / f.drive_eff + f.hotel_w
    energy_wh = bus_w * race_s / 3600
    m.update(thrust_n=p["thrust_n"], trim_deg=p["trim_deg"], prop_eff=eta_p, prop_lb=prop_mass_lb(d, f), shaft_w=shaft_w, bus_w=bus_w, energy_used_wh=energy_wh,
             motor_name=motor_spec(d).name, motor_torque_nm=mot["torque_nm"], motor_a=mot["current_a"],
             motor_v_needed=mot["volts_needed"], motor_eff=mot["efficiency"])

    # battery and voltage
    soc0 = start_soc(d.series, f)
    v_oc = d.series * ocv(soc0, cell)
    v_load = loaded_voltage(v_oc, bat["resistance_ohm"], bus_w)
    current = bus_w / v_load if v_load == v_load else float("inf")
    soc_end = soc0 - energy_wh / bat["capacity_wh"]
    v_end = loaded_voltage(d.series * ocv(max(soc_end, 0.0), cell), bat["resistance_ohm"], bus_w)
    m.update(start_soc=soc0, v_oc_start=v_oc, v_loaded=v_load, current_a=current, soc_end=soc_end,
             v_loaded_end=v_end, **prop_numbers(d, f, v, v_load))

    # limits
    spec = motor_spec(d)
    i_cap, v_cap = motor_caps(spec)
    usable_wh = f.usable_energy_fraction * bat["capacity_wh"] * soc0
    g["voltage_rule"] = (f.voltage_limit_v - v_oc, f.voltage_limit_v, "rule")
    g["motor_max_voltage"] = (v_cap - d.series * cell.v_full, v_cap, "limit")
    g["energy"] = (usable_wh - energy_wh, max(usable_wh, 1.0), "limit")
    g["shaft_power"] = (f.rating_fraction * spec.p_cont_w - shaft_w, spec.p_cont_w, "limit")
    g["battery_current"] = (f.rating_fraction * bat["current_limit_a"] - current, bat["current_limit_a"], "limit")
    g["motor_current"] = (f.rating_fraction * i_cap - mot["current_a"], i_cap, "limit")
    g["motor_voltage"] = ((v_load - mot["volts_needed"]) if v_load == v_load else -1.0, v_cap, "limit")
    g["pack_can_deliver"] = (1.0 if v_load == v_load else -1.0, 1.0, "limit")
    g["end_cell_voltage"] = ((v_end / d.series - f.min_cell_voltage_end) if v_end == v_end else -1.0, 1.0, "limit")
    g["tip_speed"] = (f.max_tip_speed_fps - m["tip_speed_fps"], f.max_tip_speed_fps, "limit")
    g["lcg_min"] = (m["lcg_frac"] - f.lcg_band[0], 0.1, "limit")
    g["lcg_max"] = (f.lcg_band[1] - m["lcg_frac"], 0.1, "limit")
    return Result(m, g)


def show(res, enforce_validity=None, title=None):
    m, p = res.m, res.m["planing"]
    if title: print(title)
    line = lambda k, v, u="": print(f"  {k:<26}{v:>10}  {u}")
    line("total mass", f"{m['mass_lb']:.1f}", f"lb   (hull {m['hull_lb']:.1f}, battery {m['battery_lb']:.1f}, motor {m['motor_lb']:.1f})")
    line("LCG", f"{100 * m['lcg_frac']:.1f}", "% of length from transom")
    line("static draft / freeboard", f"{m['static_draft_in']:.1f} / {m['freeboard_in']:.1f}", "in")
    line("GM at rest", f"{m['gmt_m']:.3f}", "m   (static roll stiffness, from the hull mesh)")
    if p["ok"]:
        line("running trim", f"{m['trim_deg']:.2f}", f"deg   wetted length/beam {p['lam']:.2f}, Cv {p['cv']:.2f}")
        line("thrust needed", f"{m['thrust_n']:.0f}", "N")
        line("shaft / bus power", f"{m['shaft_w'] / 1000:.2f} / {m['bus_w'] / 1000:.2f}", "kW")
        line("pack", f"{m['capacity_wh']:.0f}", f"Wh; start charge {100 * m['start_soc']:.0f}%, Voc {m['v_oc_start']:.1f} V, loaded {m['v_loaded']:.1f} V, {m['current_a']:.0f} A")
        line("race energy", f"{m['energy_used_wh']:.0f}", "Wh")
        line("propeller", f"{m['rpm']:.0f}", f"rpm, tip {m['tip_speed_fps']:.0f} ft/s, Kv needed {m['kv_needed']:.0f}")
        line("motor", f"{m['motor_torque_nm']:.1f}", f"N m, {m['motor_a']:.0f} A, needs {m['motor_v_needed']:.0f} V, efficiency {m['motor_eff']:.2f}  ({m['motor_name']})")
        line("propeller efficiency", f"{m['prop_eff']:.2f}", f"propeller mass {m['prop_lb']:.1f} lb")
    else:
        line("planing solver", "FAILED", p.get("error", ""))
    t = m["race_time_s"]
    line("race time", f"{int(t // 60)}:{t % 60:04.1f}", "min:sec")
    bad = res.failing(enforce_validity)
    print("  " + ("FEASIBLE" if not bad else "FAILS: " + ", ".join(bad)))
    return res
```

```python
from scipy.optimize import brentq


def tint(color, k, n):
    # n steps of one hue, light to dark, so a numeric parameter reads as ordered
    r, g, b = mpl.colors.to_rgb(color); a_ = 0.55 * (1 - k / max(n - 1, 1))
    return (r + (1 - r) * a_, g + (1 - g) * a_, b + (1 - b) * a_)


def thrust_available(d, shaft_limit_w, v_ms):
    # thrust T at which T v / eta(T) equals the shaft power the motor may deliver; eta falls as T rises, so T / eta(T) is increasing
    f_ = lambda T: T * v_ms / prop_efficiency(d, FIXED, T, v_ms) - shaft_limit_w
    return brentq(f_, 0.0, 5000.0) if f_(5000.0) > 0 else np.nan


req = series(OV, lambda r: r.m["thrust_n"])
v_load = series(OV, lambda r: r.m["v_loaded"])
shaft = series(OV, lambda r: r.m["shaft_w"])
SHOWN = {}                                         # one winding of each of these motors -> its continuous power in watts
for m_ in ("SSS 3682-V2", "SSS 4092-V2", "SSS 5684-V2", "SSS 56154-V2"):
    SHOWN[m_.replace("SSS ", "")] = next(s for s in CATALOG if s.name.startswith(m_ + " ")).p_cont_w
diams, pitches = (4, 6, 8, 10), (6, 8, 10, 14)
fig, axes = plt.subplots(2, 3, figsize=(14.5, 7.4), sharex=True, layout="constrained")

ax = axes[0, 0]                                   # force balance: what the hull needs against what the drive can deliver
ax.plot(OV_SPEEDS, req, color=INK, lw=2.5, label="needed by the hull")
for i, (label, p_cont) in enumerate(SHOWN.items()):
    c = tint(AQUA, i, len(SHOWN))
    avail = np.array([thrust_available(DESIGN, FIXED.rating_fraction * p_cont, v * MPH) for v in OV_SPEEDS])
    ax.plot(OV_SPEEDS, avail, color=c, label=f"{label} ({p_cont / 1000:g} kW) can deliver")
    held = np.where(np.isfinite(req) & (avail >= req))[0]
    if len(held): dot(ax, OV_SPEEDS[held[-1]], avail[held[-1]], c); ax.text(OV_SPEEDS[held[-1]] + 1.0, avail[held[-1]] - 24, f"{OV_SPEEDS[held[-1]]:.0f} mph", fontsize=8.5, color=INK2)
ax.set_ylim(0, 500); ax.set_ylabel("thrust (N)"); ax.set_title("Force: thrust needed against thrust available", fontsize=9.5); ax.legend(loc="upper right", fontsize=8)

ax = axes[0, 1]                                   # propeller efficiency: a bigger disk is more lightly loaded
for i, D in enumerate(diams):
    d_ = replace(DESIGN, prop_diam_in=D)
    ax.plot(OV_SPEEDS, [prop_efficiency(d_, FIXED, T, v * MPH) if np.isfinite(T) else np.nan for T, v in zip(req, OV_SPEEDS)], color=tint(BLUE, i, len(diams)), label=f"{D} in diameter")
ax.set_ylim(0, 1); ax.set_ylabel("propeller efficiency"); ax.set_title(f"Efficiency: disk loading and pitch/diameter, at pitch {DESIGN.pitch_in:g} in", fontsize=9.5); ax.legend(loc="lower right", fontsize=8)

ax = axes[0, 2]                                   # tip speed: why the diameter cannot grow without limit
for i, D in enumerate(diams):
    ax.plot(OV_SPEEDS, math.pi * D / 12 * (OV_SPEEDS * 1056 / (DESIGN.pitch_in * FIXED.slip_factor)) / 60, color=tint(BLUE, i, len(diams)), label=f"{D} in diameter")
ax.axhline(FIXED.max_tip_speed_fps, color=INK, lw=1); ax.text(OV_SPEEDS[0] + 0.5, FIXED.max_tip_speed_fps + 4, f"limit {FIXED.max_tip_speed_fps:g} ft/s", fontsize=8.5, color=INK2)
ax.set_ylabel("blade tip speed (ft/s)"); ax.set_title(f"Tip speed = pi x diameter x rpm, at pitch {DESIGN.pitch_in:g} in", fontsize=9.5); ax.legend(loc="upper left", bbox_to_anchor=(0, 0.88), fontsize=8)

ax = axes[1, 0]                                   # rpm: pitch and slip set how fast it must turn
for i, p in enumerate(pitches):
    ax.plot(OV_SPEEDS, OV_SPEEDS * 1056 / (p * FIXED.slip_factor), color=tint(ORANGE, i, len(pitches)), label=f"pitch {p} in")
ax.set_ylabel("propeller rpm"); ax.set_xlabel("speed through the water (mph)"); ax.set_title("rpm = mph x 1056 / (pitch x slip factor)", fontsize=9.5); ax.legend(loc="upper left", fontsize=8)

ax = axes[1, 1]                                   # motor drive: the winding (Kv) has to turn that rpm at the loaded pack voltage
for i, p in enumerate(pitches):
    ax.plot(OV_SPEEDS, OV_SPEEDS * 1056 / (p * FIXED.slip_factor) / v_load, color=tint(ORANGE, i, len(pitches)), label=f"pitch {p} in")
ax.set_ylabel("Kv the motor needs (rpm per volt)"); ax.set_xlabel("speed through the water (mph)"); ax.set_title("Motor drive: Kv = rpm / loaded pack voltage", fontsize=9.5)

ax = axes[1, 2]                                   # torque: power divided by angular speed, so a coarser pitch asks for more
for i, p in enumerate(pitches):
    d_ = replace(DESIGN, pitch_in=p)
    sh = np.array([T * v * MPH / prop_efficiency(d_, FIXED, T, v * MPH) if np.isfinite(T) else np.nan for T, v in zip(req, OV_SPEEDS)])
    ax.plot(OV_SPEEDS, sh / (2 * np.pi * (OV_SPEEDS * 1056 / (p * FIXED.slip_factor)) / 60), color=tint(ORANGE, i, len(pitches)), label=f"pitch {p} in")
spec_ = motor_spec(DESIGN); t_lim = 60 / (2 * np.pi * spec_.kv) * (FIXED.rating_fraction * motor_caps(spec_)[0] - spec_.i0_a)
ax.axhline(t_lim, color=INK, lw=1); ax.text(OV_SPEEDS[0] + 0.5, t_lim * 1.03, f"{spec_.name}: torque at {FIXED.rating_fraction:.0%} of max current", fontsize=8.5, color=INK2)
ax.set_ylabel("shaft torque (N m)"); ax.set_xlabel("speed through the water (mph)"); ax.set_title("Torque: shaft power / angular speed", fontsize=9.5)
headline(fig, "How propeller, thrust, motor drive and force relate: the hull sets the thrust it needs, the motor's power rating and the propeller's efficiency set the thrust available, diameter sets efficiency and tip speed, and pitch sets rpm, Kv and torque")
plt.show()
```

## 8. The fastest feasible speed

Speed is not really a free choice: for any design there is a highest speed at which every constraint still holds, and that is
the speed the boat should run. `fastest_feasible` scans speeds, keeps the feasible ones, and refines the upper edge. Race time
follows from it, so the optimizer only has to search the other variables.

```python
def fastest_feasible(d, f=FIXED, lo=8.0, hi=60.0, step=4.0, tol=0.1, enforce_validity=None):
    ok = lambda s: evaluate(replace(d, speed_mph=s), f, s)
    feas = lambda r: r.feasible(enforce_validity)
    best = None
    for s in np.arange(lo, hi + 1e-9, step):
        r = ok(s)
        if feas(r): best = (s, r)
    if best is None:
        return None
    s_lo, r_lo = best
    s_hi = s_lo + step
    if s_hi <= hi:
        while s_hi - s_lo > tol:
            mid = (s_lo + s_hi) / 2
            r = ok(mid)
            if feas(r): s_lo, r_lo = mid, r
            else: s_hi = mid
    return s_lo, r_lo


def limiting(d, f=FIXED, nudge=0.5):
    # which constraints fail just above the fastest feasible speed: what is stopping the boat
    b = fastest_feasible(d, f)
    if b is None:
        return None, []
    v, _ = b
    return v, evaluate(replace(d, speed_mph=v + nudge), f, v + nudge).failing()
```

### The problem at a glance

Everything the model is solving, on one card, computed live from the cells above: the active modes, every
decision variable against its bounds, and every constraint at this design's fastest feasible speed, sorted
with the least headroom first. A variable sitting on a bound and a margin reading `BINDING` are the two
things to look for - each is either a real optimum or a missing piece of physics. The same card is printed
for the search winner in section 14.

```python
def problem_card(d=DESIGN, f=FIXED, bar_w=24):
    """The whole problem, computed from the model itself: modes, decision variables vs BOUNDS, and
    every constraint margin at this design's fastest feasible speed, tightest first."""
    bar = lambda u: "".join("\u2588" if i == int(round(min(max(u, 0.0), 1.0) * (bar_w - 1))) else "\u00b7" for i in range(bar_w))
    b = fastest_feasible(d, f)
    v, r = b if b else (d.speed_mph, evaluate(d, f))
    print(f"PROBLEM CARD \u2014 {MODEL_VERSION}")
    print("modes:   " + "   ".join(f"{k}={v_}" for k, v_ in MODES().items()) + f"   catalog={len(CATALOG)} windings")
    print(f"objective: least race time = {f.course_miles:g} mi / ground speed + {f.allowance_s:g} s allowance, at the fastest feasible speed")
    print(f"\n{len(NAMES)} DECISION VARIABLES \u2014 this design against BOUNDS")
    print(f"{'variable':<14}{'value':>10}   {'low':>8} {'':<{bar_w}} {'high':>8}")
    for n in NAMES:
        lo, hi = BOUNDS[n]; x = getattr(d, n); u = (x - lo) / (hi - lo) if hi > lo else 0.5
        flag = ("  <- at the lower bound" if u < 0.02 else "  <- at the upper bound") if (u < 0.02 or u > 0.98) else ""
        print(f"{n:<14}{x:>10.4g}   {lo:>8g} {bar(u)} {hi:>8g}{flag}")
    where = f"at the fastest feasible speed, {v:.2f} mph" if b else f"at {v:g} mph \u2014 NO feasible speed"
    print(f"\n{len(r.margins)} CONSTRAINTS {where} \u2014 least headroom first")
    print(f"{'margin':<22}{'group':<10}{'headroom':>9}   0%{'':<{bar_w - 5}}100%+")
    for k, (mv, sc, grp) in sorted(r.margins.items(), key=lambda kv: kv[1][0] / kv[1][1]):
        u = mv / sc
        note = "  <- FAILS" if u < 0 else ("  <- BINDING" if u < 0.03 else ("  <- near" if u < 0.15 else ""))
        print(f"{k:<22}{grp:<10}{100 * u:>8.1f}%   {bar(u)}{note}")
    eq = hydrostatics(d, f, r.m["mass_lb"], r.m["vcg_m"], lcg_ft=r.m["lcg_frac"] * d.length_ft, free_trim=True)
    side = "bow" if eq["trim_deg"] > 0 else "stern"
    print(f"\nat rest: draft {eq['draft_in']:.1f} in, freeboard {eq['freeboard_in']:.1f} in, "
          f"GM {eq['gmt_m']:.2f} m, trims {abs(eq['trim_deg']):.1f} deg by the {side}")


problem_card()
```

## 11. Sweep one variable

`sweep` changes one design variable and reports, for each value, the fastest feasible speed and the race time that speed gives.
Gaps mean no speed in the scan was feasible. In the motor sweep that means the motor cannot drive this propeller (the default 10 in pitch asks for more torque than most windings can supply). This is the quickest way to see whether a variable runs to an edge and what stops it.

```python
def sweep(var, values, f=FIXED, d=DESIGN):
    """`var` may name a Design field or a Fixed field."""
    out = []
    for x in values:
        if hasattr(d, var): dd, ff = replace(d, **{var: type(getattr(d, var))(x)}), f
        else: dd, ff = d, replace(f, **{var: type(getattr(f, var))(x)})
        b = fastest_feasible(dd, ff)
        out.append((x, None if b is None else b[0], None if b is None else b[1].m["race_time_s"],
                    None if b is None else b[1].m["thrust_n"]))
    return out


def plot_sweeps(specs, f=FIXED, d=DESIGN, ncols=3):
    n = len(specs); nrows = math.ceil(n / ncols)
    fig, axes = plt.subplots(nrows, ncols, figsize=(4.3 * ncols, 3.3 * nrows), squeeze=False)
    for ax, (var, values, label) in zip(axes.flat, specs):
        res = sweep(var, values, f, d)
        xs = [r[0] for r in res]; vs = [np.nan if r[1] is None else r[1] for r in res]
        ax.plot(xs, vs, "o-", ms=3)
        ax.axvline(getattr(d if hasattr(d, var) else f, var), color="gray", ls=":", lw=1)
        ax.set_xlabel(label); ax.set_ylabel("fastest feasible speed (mph)")
        if all(np.isnan(vs)): ax.text(0.5, 0.5, "never feasible", transform=ax.transAxes, ha="center")
    for ax in axes.flat[n:]: ax.axis("off")
    plt.tight_layout(); plt.show()


plot_sweeps([
    ("beam_in", np.arange(10, 40, 1.5), "chine beam (in)"),
    ("deadrise_deg", np.arange(4, 26, 1.5), "deadrise (deg)"),
    ("length_ft", np.arange(6.5, 11.6, 0.5), "length (ft)"),
    ("depth_in", np.arange(6, 20, 1.0), "hull depth (in)"),
    ("payload_x", np.arange(0.20, 0.61, 0.025), "payload position (fraction of length)"),
    ("battery_x", np.arange(0.20, 0.61, 0.025), "battery position (fraction of length)"),
    ("motor", np.arange(0, len(CATALOG), 6), "SSS motor (catalog index, small to large)"),
    ("parallel", np.arange(1, 13, 1), "parallel strings"),
])
```

## 14. Search for a good design

Differential evolution over the design variables, with the fastest feasible speed found for each candidate (so speed is not a
variable). Series and parallel counts are integers. Pitch and diameter are searched too, because the propeller model makes them matter. Thrust angle is left out (held at zero).

This is a global search but not a proof: run it a few times with different seeds and compare. It takes about a minute.
Edit `BOUNDS` (section 2, with the inputs) to change the search space. The hull length range is the plan's; the beam range is wide on purpose so the Savitsky
validity window shows up by itself.

```python
# The search space (BOUNDS, INTEGER, NAMES) is defined with the inputs in section 2.

SEARCH_LOG = []     # one entry per candidate the last run_search evaluated:
                    # (fastest feasible speed or None, race time or penalty, failing margins at the 20 mph probe)


def design_from(x):
    kw = {n: (int(round(v)) if n in INTEGER else float(v)) for n, v in zip(NAMES, x)}
    return replace(DESIGN, **kw)


def objective(x):
    d = design_from(x)
    b = fastest_feasible(d, FIXED, step=6.0, tol=0.25)
    if b is not None:
        SEARCH_LOG.append((b[0], b[1].m["race_time_s"], ()))
        return b[1].m["race_time_s"]
    probe = evaluate(replace(d, speed_mph=20.0), FIXED, 20.0)
    SEARCH_LOG.append((None, 1000.0 + probe.violation(), tuple(probe.failing())))
    return 1000.0 + probe.violation()   # steer toward feasibility


def run_search(seed=1, popsize=8, maxiter=25):
    SEARCH_LOG.clear()
    t0 = time.time()
    r = differential_evolution(objective, [BOUNDS[n] for n in NAMES], integrality=[n in INTEGER for n in NAMES],
                               seed=seed, popsize=popsize, maxiter=maxiter, tol=1e-6, polish=False, updating="deferred")
    return r, time.time() - t0
```

```python
sr, secs = run_search(seed=1)
best_design = design_from(sr.x)
print(f"search finished in {secs:.0f} s, {sr.nfev} candidate designs, best race time {sr.fun:.1f} s")
b = fastest_feasible(best_design, FIXED)
if b:
    best_design = replace(best_design, speed_mph=b[0])     # exact, so it sits right on the limit
    show(evaluate(best_design, FIXED), title=f"Best design found, running at {best_design.speed_mph:.1f} mph")
    print("  Just above that speed these constraints fail:", limiting(best_design, FIXED)[1])
else:
    print("No feasible design was found. Try more iterations or wider bounds.")
```

### Watching the search work

The search's own record (`SEARCH_LOG`), not a summary written by hand: the best race time as the search
evaluated candidates, how many candidates were feasible as it went, and which constraint killed the ones that
never had a feasible speed (checked at a 20 mph probe). A search that stops improving early, or that wastes
most of its budget on one wall, is telling you where to look next.

```python
ok = np.array([e[0] is not None for e in SEARCH_LOG])
vals = np.array([e[1] if e[0] is not None else np.inf for e in SEARCH_LOG])
best_curve = np.minimum.accumulate(vals)
W = max(len(SEARCH_LOG) // 20, 25)
share = np.convolve(ok.astype(float), np.ones(W) / W, mode="valid")
from collections import Counter
kills = Counter(k for e in SEARCH_LOG if e[0] is None for k in e[2])

fig, axes = plt.subplots(1, 3, figsize=(13.5, 3.8), layout="constrained")
ax = axes[0]
shown = np.isfinite(best_curve)
ax.plot(np.arange(1, len(vals) + 1)[shown], best_curve[shown], color=BLUE, lw=1.8)
ax.set_xlabel("candidate designs evaluated"); ax.set_ylabel("best race time found so far (s)")
ax.set_title(f"Progress: best {best_curve[-1]:.1f} s after {len(vals)} candidates", fontsize=9.5)
ax = axes[1]
ax.plot(np.arange(len(share)) + W, 100 * share, color=ORANGE, lw=1.6)
ax.set_ylim(0, 100); ax.set_xlabel("candidate designs evaluated"); ax.set_ylabel(f"feasible share, last {W} (%)")
ax.set_title(f"Feasibility: {100 * ok.mean():.0f}% of all candidates had a feasible speed", fontsize=9.5)
ax = axes[2]
top = kills.most_common(8)[::-1]
ax.barh([k for k, _ in top], [c for _, c in top], color=RED, height=0.6)
ax.set_xlabel("infeasible candidates failing it (at 20 mph)"); ax.grid(axis="y", visible=False)
ax.set_title(f"What killed the {int((~ok).sum())} candidates with no feasible speed", fontsize=9.5)
plt.show()
```

### Why the search chose this design

The cell below takes the best design and says why, using the model's own numbers: which constraints it is up against, how hard the motor is working
(against its own limit and against the strongest winding in the catalog), and what happens if the same boat is given every other winding. Run it after
any search; it explains whatever the search returned.

```python
# ---- Why the search chose this design ----
from collections import Counter


def torque_ceiling(s, f=FIXED):
    # most torque a winding may deliver: Kt x (usable fraction of max current - no-load current)
    return 60 / (2 * math.pi * s.kv) * (f.rating_fraction * motor_caps(s)[0] - s.i0_a)


def explain_best(d, f=FIXED):
    r = evaluate(d, f)
    m, spec = r.m, motor_spec(d)
    print(f"Why this design, running at {d.speed_mph:.1f} mph")

    # 1. what stops it: the margins closest to zero (voltage_rule is zero by construction, so it is skipped)
    tight = sorted((v / s, k) for k, (v, s, g) in r.margins.items() if k != "voltage_rule")[:3]
    print("\n1. What stops it going faster (margin as a fraction of its scale; 0 means right at the limit):")
    for ratio, k in tight:
        print(f"     {k:<20}{ratio:7.3f}")

    # 2. how hard the motor is working, against this winding and against the strongest winding in the catalog
    cap = torque_ceiling(spec, f)
    top_cap, top = max((torque_ceiling(s, f), s.name) for s in CATALOG)
    used = m["motor_torque_nm"] / cap
    print(f"\n2. The motor: {spec.name} ({spec.kv:g} Kv, {motor_caps(spec)[0]:g} A cap, {MOTOR_LIMITS} limits)")
    print(f"     torque asked for      {m['motor_torque_nm']:5.2f} N m at {m['rpm']:.0f} rpm")
    print(f"     this winding's limit  {cap:5.2f} N m at {f.rating_fraction:.0%} of its max current ({100 * used:.0f}% used; {m['motor_a']:.0f} of {f.rating_fraction * motor_caps(spec)[0]:.0f} A)")
    print(f"     strongest in catalog  {top_cap:5.2f} N m ({top})")
    print(f"     voltage               needs {m['motor_v_needed']:.0f} V of the {m['v_loaded']:.0f} V the loaded pack gives; efficiency {m['motor_eff']:.2f}")
    if m["motor_torque_nm"] > 0.95 * top_cap:
        print("     -> The boat is at the torque ceiling of the strongest motor in the catalog. The motor, not the hull, sets the speed.")
    elif used > 0.97:
        print("     -> This winding is at its own current limit. A stronger winding or a lower-torque propeller (finer pitch) would help.")
    else:
        print("     -> The motor has room; something else sets the speed.")

    # 3. the same boat with every other winding in the catalog
    rows, why = [], Counter()
    for i, s in enumerate(CATALOG):
        dd = replace(d, motor=i)
        b = fastest_feasible(dd, f, step=6.0, tol=0.25)
        if b: rows.append((b[0], s))
        else: why.update(evaluate(dd, f).failing())
    rows.sort(key=lambda t: -t[0])
    print(f"\n3. The same boat with each of the {len(CATALOG)} windings: {len(rows)} can drive it at some speed. The fastest:")
    for v_, s in rows[:6]:
        print(f"     {s.name:<18}{v_:5.1f} mph   {s.kv:g} Kv, {motor_caps(s)[0]:g} A, ceiling {torque_ceiling(s, f):.1f} N m, {s.mass_lb:.1f} lb")
    if why:
        print("   The rest fail on (windings that fail each check at this speed):", ", ".join(f"{k} {n}" for k, n in why.most_common(3)))
    print("\n   Reading it: windings of one motor share almost the same torque ceiling (Kv x max current is nearly constant), so they tie when torque is the limit."
          "\n   Kv then decides which one suits: a low-Kv winding makes more torque per amp, so it can carry a coarse, slow propeller within its current limit, but it needs"
          "\n   more pack voltage per rpm (rpm / Kv + I x R). The search also weighs the motor's weight and the hull's own trim limit.")

explain_best(best_design)
```

## Approaches

The same design space (`BOUNDS`) searched four ways with the same budget of about 2,000 evaluated designs, and with three
different objectives. The first chart plots the best race time found against the number of designs evaluated; the second shows
where each result puts every variable inside its range, with the edges shaded.

Things to try: change `SEEDS`, change `TARGET_MPH`, widen `BOUNDS` in section 14, or add a method to `run_method`.

```python
# --- Approaches: the same design space searched different ways -------------------------------------------------
class Tracked:
    # wraps an objective so every method is judged on the same thing: the best value after each evaluation
    def __init__(self, fn):
        self.fn, self.best, self.best_x, self.trace = fn, float("inf"), None, []

    def __call__(self, x):
        v = float(self.fn(x))
        if v < self.best: self.best, self.best_x = v, np.array(x, dtype=float)
        self.trace.append(self.best)
        return v


def obj_race_time(x):
    return objective(x)                                    # section 14's objective: race time at the fastest feasible speed


def _at_target(x):
    d = replace(design_from(x), speed_mph=TARGET_MPH)
    return d, evaluate(d, FIXED, TARGET_MPH)


def obj_lightest(x):
    d, r = _at_target(x)                                   # least total mass among boats that can run TARGET_MPH
    return r.m["mass_lb"] if r.feasible() else 1000.0 + 100 * r.violation()


def obj_least_energy(x):
    d, r = _at_target(x)                                   # least race energy among boats that can run TARGET_MPH
    return r.m["energy_used_wh"] if r.feasible() else 1000.0 + 100 * r.violation()


def run_method(method, fn, seed=0, budget=8 * len(NAMES) * 26):          # the budget differential evolution uses (popsize 8, 25 generations)
    from scipy.optimize import dual_annealing, minimize
    t, bounds = Tracked(fn), [BOUNDS[n] for n in NAMES]
    t0 = time.time()
    if method == "de":
        differential_evolution(t, bounds, integrality=[n in INTEGER for n in NAMES], seed=seed, popsize=8, maxiter=25,
                               tol=1e-6, polish=False, updating="deferred")
    elif method == "random":
        rng = np.random.default_rng(seed)
        for _ in range(budget): t(np.array([rng.uniform(*b) for b in bounds]))
    elif method == "annealing":
        dual_annealing(t, bounds, seed=seed, maxfun=budget)
    elif method == "powell":
        minimize(t, np.array([(lo + hi) / 2 for lo, hi in bounds]), method="Powell", bounds=bounds, options=dict(maxfev=budget))
    return dict(x=t.best_x, value=t.best, trace=t.trace, nfev=len(t.trace), secs=time.time() - t0)
```

```python
SPECS = ([(f"DE seed {s}", "de", obj_race_time, s, "race time") for s in SEEDS]
         + [("Random search", "random", obj_race_time, 0, "race time"), ("Dual annealing", "annealing", obj_race_time, 0, "race time"),
            ("Powell (local)", "powell", obj_race_time, 0, "race time"),
            (f"Lightest at {TARGET_MPH:g} mph", "de", obj_lightest, 1, "lightest"), (f"Least energy at {TARGET_MPH:g} mph", "de", obj_least_energy, 1, "least energy")])
RUNS = []
for label, method, fn, seed, goal in SPECS:
    out = run_method(method, fn, seed)
    b = fastest_feasible(design_from(out["x"]), FIXED) if out["x"] is not None else None
    if b is None:
        print(f"{label}: no feasible design found"); continue
    d, r = replace(design_from(out["x"]), speed_mph=b[0]), b[1]
    RUNS.append(dict(label=label, goal=goal, method=method, design=d, result=r, **out))

DESIGNS = {run["label"]: (run["design"], run["result"]) for run in RUNS}
for name, d in EXTRA_DESIGNS.items():
    b = fastest_feasible(d, FIXED)
    if b: DESIGNS[name] = (replace(d, speed_mph=b[0]), b[1])
    else: print(f"{name}: no feasible speed")
DNAMES = list(DESIGNS)
DCOLOR = {n: CAT[i % len(CAT)] for i, n in enumerate(DNAMES)}
if SUBJECT is None:
    SUBJECT = min((run for run in RUNS if run["goal"] == "race time"), key=lambda run: run["value"])["label"]
if COMPARE is None:
    COMPARE = [SUBJECT] + [n for n in (f"Lightest at {TARGET_MPH:g} mph", f"Least energy at {TARGET_MPH:g} mph", "Random search") if n in DESIGNS and n != SUBJECT]
    COMPARE = (COMPARE + [n for n in DNAMES if n not in COMPARE])[:4]
SUBJ, SUBJ_R = DESIGNS[SUBJECT]

print(f"{'design':<26}{'objective':>10}{'evals':>7}{'secs':>6}   {'mph':>5} {'race s':>7} {'lb':>6} {'Wh':>6}  {'motor':<18} S x P")
for run in RUNS:
    d, r = run["design"], run["result"]
    print(f"{run['label']:<26}{run['value']:>10.1f}{run['nfev']:>7}{run['secs']:>6.0f}   {d.speed_mph:>5.1f} {r.m['race_time_s']:>7.0f} "
          f"{r.m['mass_lb']:>6.1f} {r.m['capacity_wh']:>6.0f}  {motor_spec(d).name[4:]:<18} {d.series}S{d.parallel}P")
print(f"\nsubject: {SUBJECT}   overlaid in line charts: {COMPARE}")
```

```python
def utilization(d, r):
    # each limit as a share of its allowed value; above 100% the limit is broken
    m, c = r.m, FIXED.cell
    cap_a = FIXED.rating_fraction * d.parallel * c.max_current_a
    s = motor_spec(d)
    motor_a = FIXED.rating_fraction * motor_caps(s)[0]
    usable = FIXED.usable_energy_fraction * m["capacity_wh"] * m["start_soc"]
    return dict(energy=100 * m["energy_used_wh"] / usable, shaft=100 * m["shaft_w"] / (FIXED.rating_fraction * s.p_cont_w),
                pack=100 * m["current_a"] / cap_a, motor=100 * m["motor_a"] / motor_a)


fig, axes = plt.subplots(1, 4, figsize=(13, 3.4), sharey=True, layout="constrained")
labels = dict(energy="energy used / usable energy", shaft="shaft power / 80% of motor rating",
              pack="battery current / 80% of cell limit", motor="motor current / 80% of its max current")
for n in COMPARE:
    d = DESIGNS[n][0]
    u = [utilization(d, r) if r.m["planing"]["ok"] and r.m.get("v_loaded") == r.m.get("v_loaded") else None for r in CURVES[n]]
    for ax, key in zip(axes, labels):
        ax.plot(SPEED_SCAN, [np.nan if x is None else x[key] for x in u], color=DCOLOR[n], label=n)
        dot(ax, d.speed_mph, utilization(d, DESIGNS[n][1])[key], DCOLOR[n], s=34)
for ax, key in zip(axes, labels):
    ax.axhline(100, color=INK, lw=1); ax.set_title(labels[key], fontsize=9); ax.set_xlabel("speed (mph)"); ax.set_ylim(0, 160)
axes[0].set_ylabel("percent of the limit"); axes[0].legend(loc="upper left", fontsize=8)
headline(fig, "Each power and energy limit as a share of what it allows; dots mark each design's fastest feasible speed")
plt.show()
```

## 17. What to replace next

- **Cell data** (`Cell`): use a real datasheet, especially resistance, continuous current and the voltage-versus-charge curve.
- **Motor data** (`sss_motors.csv`): typed from the supplier sheet, and 33 windings with a doubtful resistance entry are left out. Confirm those
  rows with the supplier, or measure the motor you buy. The no-load current is the sheet's 11.1 V figure held constant, and there is no gearbox.
- **Speed controller** (`drive_eff`): a placeholder 97%.
- **Propeller** (`prop_blade_eff`, `prop_pd_opt`, `prop_pd_penalty`, `prop_mass_ref_lb`, `max_tip_speed_fps`): the structure is actuator-disk
  theory, but the blade losses, the pitch/diameter penalty, the mass law and the tip-speed limit are placeholders. Replace them with a
  propeller map (Wageningen B-series or the manufacturer's curves) once a propeller is chosen.
- **Prismatic coefficient** (`prismatic_coeff`): sets static buoyancy and therefore freeboard.
- **Wave height and current** (`wave_height_in`, `current_mph`): set from the race venue.
- **Placement heights and gyradius**: from the real layout.
- **Deferred hull variables** (transom beam, bow deadrise, keel curve) are not in this notebook; OpenPlaning cannot use them.
